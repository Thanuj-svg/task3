```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}
# Step 1: Find highest score
high = max(employees.values())
# Step 2: Find all employees with highest score
top = []
for name, score in employees.items():
    if score == high:
        top.append(name)
print("Top Performers Eligible for Bonus:", ", ".join(top), f"(Score: {high})")
```

```python
text = "Buy mobile phone buy phone online"
# Step 1: convert to lowercase
text = text.lower()
# Step 2: split into words
words = text.split()
# Step 3: count frequency using dictionary
feq = {}
for word in words:
    if word in feq:
        feq[word] += 1
    else:
        feq[word] = 1
# Step 4: print only repeated keywords
result = {}
for key, value in feq.items():
    if value > 1:
        result[key]=value
print(result)
```

```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]
domain_count = {}
# Count domains
for email in emails:
    domain = email.split("@")[1]   # get domain part

    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1
# Calculate percentage
total = len(emails)
for domain, count in domain_count.items():
    percentage = (count / total) * 100
    print(f"{domain}: {int(percentage)}%")
```

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
# Step 1: Calculate average
average = sum(sales) / len(sales)
# Step 2: Threshold (30% above average)
threshold = average * 1.3
# Step 3: Find spike days
for i in range(len(sales)):
    if sales[i] > threshold:
        print(f"Day {i+1}: {sales[i]}")
```

```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
counts = {}
# Count occurrences
for user in user_ids:
    if user in counts:
        counts[user] += 1
    else:
        counts[user] = 1
# Print only duplicates
for user, count in counts.items():
    if count > 1:
        print(f"{user} -> {count} times")
```
